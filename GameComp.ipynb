Number Guessing Game

```python
from random import randint
from os import system

print("Welcome to the Number Guessing Game!")
numguess = randint(1, 100)

while True:
    att_input = input("How many attempts would you like to have? ")
    try:
        att = int(att_input)
    except ValueError:
        print("Please enter a valid integer for attempts.")
        continue
    if att <= 0:
        print("Please enter a positive number of attempts.")
        continue
    break

print("I have selected a number between 1 and 100. Can you guess it?")
attempts_left = att

while attempts_left > 0:
    guess_input = input("Enter your guess: ")
    try:
        guess = int(guess_input)
    except ValueError:
        print("Please enter a valid integer.")
        continue
    if guess < 1 or guess > 100:
        print("Your guess is out of bounds. Please guess a number between 1 and 100.")
        continue

    attempts_left -= 1

    if guess < numguess:
        print("Too low! Try again.")
        system('clear')
    elif guess > numguess:
        print("Too high! Try again.")
        system('clear')
    else:
        print(f"Congratulations! You guessed the number {numguess}.")
        print(f"Attempts remaining: {attempts_left}")
        break
else:
    print(f"Sorry, you've run out of attempts. The number was {numguess}.")
```

Survival Game (original idea by Cobble918)

```python
from random import randint as ri


def randay():
    roll = ri(1, 10)

    if roll == 1:
        return "Food Loss"
    if roll == 2:
        return "Storm"
    return None


def monster():
    roll = ri(1, 5)
    monsters = {
        1: ("Wolf", ri(5, 10), ri(2, 4), ri(0, 2)),
        2: ("Hunter", ri(8, 12), ri(3, 6), ri(1, 3)),
        3: ("Bear", ri(12, 18), ri(4, 7), ri(2, 4)),
        4: ("Adventurer", ri(10, 14), ri(5, 8), ri(2, 4)),
        5: ("Soldier", ri(14, 16), ri(6, 9), ri(3, 5)),
    }
    name, health, damage, defence = monsters[roll]
    return {
        "name": name,
        "health": health,
        "damage": damage,
        "defence": defence,
    }


player = {
    "health": 10,
    "money": 1000,
    "food": 10,
    "damage": 10,
    "defence": 5,
}

print("Health:", player["health"])
print("Money:", player["money"])
print("Food:", player["food"])
print("Damage:", player["damage"])
print("Defence:", player["defence"])
print("")
print("You will use these resources to survive on a deserted island.")
print("Every night, monsters will attack you and you will fight or flee them")
print("Every night is different")
print("Good luck!")
print("")
print("===============================GAME START===============================")
night = 1

Disaster = randay()
Mon = monster()

print("Today's disaster is", Disaster or "None")
if Disaster == "Food Loss":
    food_loss = ri(1, 5)
    player["food"] = max(0, player["food"] - food_loss)
    print("You lost some food due to spoilage!")
    print("Food remaining:", player["food"])
elif Disaster == "Storm":
    money_loss = ri(25, 200)
    food_loss = ri(1, 3)
    dmg_loss = ri(1, 3)
    player["money"] = max(0, player["money"] - money_loss)
    player["food"] = max(0, player["food"] - food_loss)
    player["damage"] = max(1, player["damage"] - dmg_loss)
    print("A storm damaged your supplies and washed away some of your money!")
print("")
print("Night", night)
print("A monster approaches!")
print(f"It's a {Mon['name']}!")
print(
    f"These {Mon['name']}s have {Mon['health']} health, deal {Mon['damage']} damage, "
    f"and have {Mon['defence']} defence."
)

action = input(
    f"Do you want to attack and deal up to {max(0, player['damage'] - Mon['defence'])} damage "
    f"or flee and maybe take up to {max(0, Mon['damage'] - player['defence'])} damage? (A/F) "
).strip().lower()

if action == 'a':
    damage_to_monster = max(0, player['damage'] - Mon['defence'])
    Mon['health'] -= damage_to_monster
    print(f"You strike the {Mon['name']} for {damage_to_monster} damage!")
    if Mon['health'] <= 0:
        print(f"The {Mon['name']} is defeated!")
    else:
        damage_taken = max(0, Mon['damage'] - player['defence'])
        player['health'] -= damage_taken
        print(f"The {Mon['name']} counterattacks for {damage_taken} damage. Health: {player['health']}")
else:
    damage_taken = max(0, Mon['damage'] - player['defence'] - ri(0, 2))
    player['health'] -= damage_taken
    print(f"You flee but take {damage_taken} damage while escaping. Health: {player['health']}")

print("")
print("=== End of Night ===")
print(
    f"Health: {player['health']} | Money: {player['money']} | Food: {player['food']} | Damage: {player['damage']}"
)
```
